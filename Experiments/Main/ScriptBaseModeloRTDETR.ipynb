```python
# Load a COCO-pretrained RT-DETR model
model = RTDETR("nombre_modelo") 

# Display model information (optional)
model.info()
```

```python
# Entrenar
model.train(
    data="ruta_data.yaml",
    epochs=200,
    patience=20,
    imgsz=640,
    batch=8,
)
```

```python
#Evaluamos el modelo entrenado
metrics = model.val(data="ruta_data.yaml", plots=True)

# Mostrar resultados
print(metrics)
```
