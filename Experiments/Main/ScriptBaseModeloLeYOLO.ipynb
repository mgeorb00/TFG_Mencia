```python
import sys
import os
import torch



# 3. Importar la librería local libre de conflictos globales
from ultralytics import YOLO
print("¡LeYOLO importado correctamente!")

# 4. Cargar los pesos preentrenados usando el parche
model = YOLO("ruta_modelo.pt")
print("¡Pesos LeYOLOMedium.pt cargados con éxito!")

# 5. Iniciar el entrenamiento en la GPU T4 de Colab usando tu YAML
print("Iniciando entrenamiento en GPU...")
results = model.train(
    data='ruta_data.yaml', # Tu archivo YAML actualizado
    epochs=200,
    imgsz=640,
    batch=8,                       # Si la GPU da error "Out of memory", cámbialo a 8
    device=0,
    patience= 20
)
```

```python
import sys
import os
import torch

path_best_weights = "ruta_best.pt"
model = YOLO(path_best_weights)

print("Iniciando validación del modelo entrenado...")

# 4. Ejecutar el modo de validación básica
metrics = model.val(
    data='ruta_data.yaml', # Tu archivo YAML
    imgsz=640,
    batch=16,
    device=0,                                                    # Usar GPU
    split='val'                                                  # Evaluar sobre el conjunto de validación
)

print("\n--- RESULTADOS DE LA VALIDACIÓN ---")
print(f"mAP50: {metrics.box.map50:.4f}")
print(f"mAP50-95: {metrics.box.map:.4f}")
print(f"Precisión (Precision): {metrics.box.mp:.4f}")
print(f"Sensibilidad (Recall): {metrics.box.mr:.4f}")
```
