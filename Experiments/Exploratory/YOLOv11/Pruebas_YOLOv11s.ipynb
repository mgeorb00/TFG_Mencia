```python
from ultralytics import YOLO

# Cargar el modelo preentrenado
model = YOLO('yolo11s.pt')

# Entrenar
model.train(
     #data='coco8.yaml',
    data='coco128.yaml',  
    epochs=3,
    imgsz=640
)
```

```python
# Evaluar sobre el dataset COCO8
#metrics = model.val(data="coco8.yaml", plots=True)
metrics = model.val(data="coco128.yaml", plots=True)

# Mostrar resultados 
print(metrics)
```

```python
import time
# Cargar una imagen o video
#source = "datasets/coco8/images/val/000000000036.jpg"
source = "datasets/coco128/images/val/000000000009.jpg"

# Medir tiempo de inferencia
t0 = time.time()
results = model(source)
t1 = time.time()

elapsed = t1 - t0
fps = 1 / elapsed
print(f"Inferencia en {elapsed:.3f}s → {fps:.2f} FPS")
```

```python
#--------------------> Utilizado para comprobar las metricas utilizando coco8, se entrena el modelo YOLO con COCO8

from ultralytics import YOLO
import time

# -----------------------------
# CONFIGURACIÓN
# -----------------------------
MODEL_PATH = "runs/detect/train13/weights/best.pt"  # Ruta a tu modelo entrenado
DATA_YAML = "coco128.yaml"                           # Ruta a tu dataset YAML
EPOCHS = 50                                        # Número de épocas usadas en el entrenamiento
FPS_ITERATIONS = 10                                # Número de veces que se medirá FPS para promedio
TEST_IMAGE = "datasets/coco128/images/val/000000000009.jpg"  # Imagen de prueba para FPS

# # -----------------------------
# # CARGAR MODELO Y VALIDAR
# # -----------------------------
# model = YOLO(MODEL_PATH)
# metrics = model.val(data=DATA_YAML)

# -----------------------------
# INFO DEL MODELO
# -----------------------------
layers, params, grads, gflops = model.info()
params_m = params / 1e6

# -----------------------------
# MÉTRICAS PRINCIPALES
# -----------------------------
res = metrics.results_dict
ap50 = res['metrics/mAP50(B)']
ap50_95 = res['metrics/mAP50-95(B)']
precision = res['metrics/precision(B)']
recall = res['metrics/recall(B)']

# Calcular F1-score y “Accuracy” (aproximado)
f1_score = 2 * (precision * recall) / (precision + recall)
accuracy = (precision + recall) / 2  # No es accuracy real, pero sirve como métrica representativa

# -----------------------------
# FPS (Frames por segundo)
# -----------------------------
t_total = 0
for _ in range(FPS_ITERATIONS):
    t0 = time.time()
    _ = model(TEST_IMAGE)
    t_total += (time.time() - t0)
fps = FPS_ITERATIONS / t_total

# -----------------------------
# IMPRIMIR RESULTADOS
# -----------------------------
print("\n========== MÉTRICAS DEL MODELO ==========")
print(f"Epochs: {EPOCHS}")
print(f"Params(M): {params_m:.2f}")
print(f"GFLOPs: {gflops}")
print(f"FPS (bs=1): {fps:.2f}")
print(f"Precision: {precision:.3f}")
print(f"Recall: {recall:.3f}")
print(f"Accuracy (aprox): {accuracy:.3f}")
print(f"F1-score: {f1_score:.3f}")
print(f"APval50: {ap50:.3f}")
print(f"APval50-95: {ap50_95:.3f}")
print("========================================\n")
```
